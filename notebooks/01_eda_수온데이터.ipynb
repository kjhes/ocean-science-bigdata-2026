# 01. 지역별 수온 데이터 탐색적 분석 (EDA)

원자료는 `수온 데이터/{지역(관측소)}/*.csv`입니다. 저장소 루트 또는 notebooks 폴더에서 실행합니다.
EDA는 3일 이하 내부 결측만 보간하며 긴 결측은 남깁니다. 보간된 표를 예측 평가에 사용하지 않습니다.
예측 실행: `python src/models/run_forecast.py` (저장소 루트).

```python
import sys
from pathlib import Path
root = Path.cwd() if (Path.cwd() / 'src').exists() else Path.cwd().parent
sys.path.insert(0, str(root / 'src'))
from data.load_data import load_all_regions
from features.preprocess import preprocess_pipeline, region_summary
from visualization.plots import plot_region_trend, plot_monthly_boxplot, plot_region_comparison_bar
```

## 1. 원자료 로드

```python
raw_df = load_all_regions()
raw_df = raw_df[raw_df['date'].between('2021-01-01', '2025-12-31')].copy()
raw_df.head()
```

## 2. 전처리 (결측치/이상치 처리, 정렬, 파생변수 추가)

```python
df = preprocess_pipeline(raw_df)
df.head()
```

## 3. 지역별 기초 통계

```python
summary = region_summary(df)
summary
```

## 4. 지역별 수온 변화 추이 시각화

```python
plot_region_trend(df, save=True)
```

```python
plot_region_comparison_bar(summary, save=True)
```

## 5. 월별 분포 (계절성 확인) - 지역별로 반복

```python
for region in df['region'].unique():
    plot_monthly_boxplot(df, region, save=True)
```
